```python
import numpy as np
```

```python
#### MARKS OF STUDENTS
## Rows -> Students , Columns -> Maths, Science, English
marks = np.array([[78,85,88],
                  [90,92,95],
                  [45,52,60],
                  [30,40,35],
                  [66,70,72]])

#### ATTENDANCE PERCENTAGE
attendance = np.array([85,90,60,50,75])

students = np.array(["Pooja", "Tejal", "Minu", "Jatin", "Pankaj"])
```

```python
#### AVERAGE MARKS OF EACH STUDENT
avg_marks = np.mean(marks, axis=1)

print("Average Marks: ")
for i in range(len(students)):
    print(students[i], ":", avg_marks[i])
```

```python
#### SUBJECT WISE HIGHEST AND LOWEST MARKS
subjects = ["Maths", "Science", "English"]

print("\nHIGHEST Marks in each subject: ")
print(np.max(marks, axis=0))

print("\nLOWEST Marks in each subject: ")
print(np.min(marks, axis=0))
```

```python
#### CLASS TOPPER
topper_index = np.argmax(avg_marks)
print("\n Class Topper: ", students[topper_index])
```

```python
#### STUDENTS WHO FAILED (AVG < 40)
failed = students[avg_marks < 40]
print("\n Failed Students: ", failed)
```

```python
#### ATTENDANCE CHECK
low_attendance = students[attendance < 75]
print("\n Low Attendance Students: ", low_attendance)
```

```python
#### GRADE SYSTEM
grades = []

for avg in avg_marks:
    if avg >= 80:
        grades.append("A")
    if avg >= 60:
        grades.append("B")
    if avg >= 40:
        grades.append("C")
    else:
        grades.append("Fail")

print("\nGRADES: ")

for i in range(len(students)):
    print(students[i], ":", grades[i])
```
